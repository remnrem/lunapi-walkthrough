## Simple tabulations

Luna's `ANNOTS` command can be used to generate simple statistics (e.g. counts, durations) for the annotations associated with a recording:

```python
# luna harm1.lst -o out.db -s ANNOTS
res = proj.silent_proc( 'ANNOTS' ) 
```

```python
# destrat out.db +ANNOTS -r ANNOT INST T > tmp/ann.1
tbl = proj.table( 'ANNOTS' , 'ANNOT_INST_T1_T2' ) 
tbl
```

Counting the total number of annotations: 

```python
# table( d$ANNOT )
tbl.ANNOT.value_counts().sort_index()
```

We can extract the counts stratified by individual:

```python
# table( d$ID, d$ANNOT )
import pandas as pd
pd.crosstab( tbl.ID , tbl.ANNOT ) 
```

Finally, we can look at the average duration of each annotation class:

```python
# round( tapply( d$DUR , list( d$ID , d$ANNOT ) , mean ) ,2 ) 
tbl.groupby( [ 'ID', 'ANNOT' ] )[ 'DUR' ].mean().unstack().reset_index()
```
