## Initiate lunapi

```python
import lunapi as lp
proj = lp.proj()
```

## Build a sample list

Assuming all data are in `../work/data` we can get a list of the files as follows: 

```python
import os
os.listdir('../work')
```

Note that with the exception of the initial index, this and all other notebooks are in the `p1/` or `p2/`, etc,  folders and so we need to use `..` to point to the parent directory.

Build an initial sample list.  Based on reviewing the file extensions of the annotation files, we'll explicitly specify them (i.e. so that `build` knows how to match `file1.edf` with `file1.tsv` etc.  This should return the number of EDF files found (i.e. 20).

```python
proj.build( [ '../work/data/' , '-ext=csv,annot,eannot,xml,tsv' ] )
```

Display the sample list:

```python
proj.sample_list()
```

Noting that one file was mislabeled, and so it is missing in the above list.  We can rename `F04b.annot` to `F04.annot`:

```python
os.rename('../work/data/annots/F04b.annot', '../work/data/annots/F04.annot')
```

We'll now rebuild the sample-list - this now attaches all N=20 individuals correctly:

```python
proj.build( [ '../work/data/' , '-ext=csv,annot,eannot,xml,tsv' ] ) 
proj.sample_list()
```

## Save the sample list

To facilitate working with the other notebooks we'll save the sample list as a text file. This means we'll be able to load it easily again -- although in principle one can also easily `build()` the lunapi sample list instead, as above.  This is more to show how the various components of Luna can be used together.    

In this particular case, we'll need to strip out some special characters from the annotations field (the `{` and `}` characters above (as the annotation column reflects a _set_ of annotation files, not just one).

```python
sl = proj.sample_list()
sl['Annotations'] = sl['Annotations'].map( lambda x: str(x).strip('{}\'') )
sl
```

Write the sample list to a file `s1.lst` in the parent folder:

```python
sl.to_csv( '../s1.lst', sep="\t", index=None, header=False)
```

We could have achieved the same thing (building `s1.lst`) by hand, by saving a spreadsheet as a tab-delimited text file, or by using the command-line `luna --build` option, as in the primary walk-through.
